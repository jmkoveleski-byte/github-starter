### 1.0. Budget Comparison Challenge

You're organizing an event. Your current budget is $1,000.

- You expect 75 attendees.
- Each attendee costs $12.25 to serve (including food and materials).
- Calculate the **total cost** of the event.
- Compare the total cost to your budget.
- Print a message that says whether you're **within budget** or **over budget**.

➡️ Use your result to answer the quiz question: Are you over or under budget?

````python
attendees = 75
cost_per_person = 12.25
budget = 1000

# Calculate total cost
total_cost = attendees * cost_per_person
print(f"Total Cost: ${total_cost:.2f}")

# Compare to budget and print result
if total_cost <= budget:
    print("Within budget")
else:
    print("Over budget")
````

### 1.2. Pizza Cost per Square Inch Comparison

You're deciding between two pizzas:

- A 13-inch pizza that costs $18
- A 15-inch pizza that costs $20.50

Using the formula for the area of a circle $A = \pi r^2$, compute the **cost per square inch** for both pizzas. Use the equation: **cost per square inch = total cost ÷ area of the pizza.**

➡️ Use your result to answer the quiz: Which pizza is cheaper per square inch?

````python
# Use 3.14159 for π
pi = 3.14159

# Calculate and compare the cost per square inch for both pizzas

#13 Inch Pizza 1
cost_13 = 18
radius_13 = 13 / 2
area_13 = pi * radius_13 ** 2
cost_per_sq_inch_13 = cost_13 / area_13

#pizza 2
cost_15 = 20.50
radius_15 = 15 / 2
area_15 = pi * radius_15 ** 2
cost_per_sq_inch_15 = cost_15 / area_15

# Print the results
print(f"The 13 inch Pizza Cost per Square Inch: ${cost_per_sq_inch_13:.2f}")
print(f"The 15 inch Pizza Cost per Square Inch: ${cost_per_sq_inch_15:.2f}")

if cost_per_sq_inch_13 < cost_per_sq_inch_15:
    print("13 inch Pizza is cheaper per square inch.")
else:
  print("15 inch Pizza is cheaper per square inch.")
````

### 3.1. Days Until Christmas

Use Python's `datetime` module to calculate how many days are between **July 4, 2026** and **December 25, 2026**.

➡️ The quiz will ask how many days are between those two dates.

````python
from datetime import date

# Your code here
christmas_date = date(2026, 12, 25)
today = date(2026,7,4)

# Calculate the number of days until Christmas
days_until_christmas = (christmas_date - today).days
print(f"Days until Christmas: {days_until_christmas} days")
````
